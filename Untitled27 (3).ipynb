```python
import numpy as np
import pandas as pd

# Example DataFrame
# df = pd.read_csv('your_data.csv')

# List of columns to be excluded from log‐transform consideration
exclude_cols = [f'f{i}' for i in range(226, 310)]

log_transform_cols = []

# Iterate through numerical columns, skipping excluded ones
for col in df.select_dtypes(include=[np.number]).columns:
    if col in exclude_cols:
        continue
    
    # Ensure column has no negative values before considering log transform
    if (df[col] < 0).any():
        continue

    mean_val = df[col].mean()
    median_val = df[col].median()
    
    # Condition 1: median == 0 but some positive mass → skewed
    if median_val == 0 and mean_val > 0.1:
        log_transform_cols.append(col)
        
    # Condition 2: non-zero median with mean/median ratio
    elif pd.notnull(mean_val) and pd.notnull(median_val) and median_val != 0:
        ratio = mean_val / median_val
        if ratio > 2.0:
            log_transform_cols.append(col)

# Apply log1p and drop originals
for col in log_transform_cols:
    df[f'{col}_log'] = np.log1p(df[col])

df.drop(columns=log_transform_cols, inplace=True)

print(f"Transformed and dropped {len(log_transform_cols)} columns:")
print(log_transform_cols)
```

```python
import pandas as pd
import numpy as np

# === 1. Expand Range Helper ===
def expand_range(prefix, start, end):
    return [f"{prefix}{i}" for i in range(start, end + 1)]

# === 2. Feature Lists ===
impute_with_zero_cols = (
    # All behavioral counts, flags, spend, time, impressions, and OHE features
    ['f13'] + expand_range('f', 14, 27) +
    ['f28', 'f29', 'f30', 'f31', 'f32', 'f33', 'f34', 'f35', 'f36', 'f37', 'f38'] + # Added f38
    expand_range('f', 94, 101) + expand_range('f', 104, 111) + ['f113'] +
    expand_range('f', 123, 129) + expand_range('f', 130, 134) + ['f137', 'f138'] +
    expand_range('f', 140, 145) +
    expand_range('f', 146, 151) + expand_range('f', 199, 204) +
    ['f205'] + expand_range('f', 206, 213) +
    ['f217', 'f218', 'f219', 'f220', 'f221', 'f222'] +
    expand_range('f', 226, 309) + # Added all One-Hot Encoded Features
    expand_range('f', 310, 324) + expand_range('f', 325, 333) + # Added redemption modes
    expand_range('f', 336, 340) + ['f344', 'f345', 'f346', 'f348'] +
    expand_range('f', 361, 366) +
    ['f39', 'f40', 'f41'] + expand_range('f', 152, 186) +
    expand_range('f', 59, 76)
)

impute_with_median_or_mode_cols = (
    # All scores, profile attributes, and ratio features
    expand_range('f', 1, 12) +
    ['f42'] + expand_range('f', 43, 58) +
    expand_range('f', 77, 79) + ['f81', 'f82', 'f83', 'f85'] +
    expand_range('f', 86, 87) + expand_range('f', 89, 91) + ['f93'] + ['f187','f188','f190','f191','f192','f193','f194','f195','f196','f197','f198']+
    expand_range('f', 114, 119) + ['f121'] + ['f139'] +
    ['f214', 'f215', 'f216'] + expand_range('f', 223, 225) +
    ['f341', 'f342', 'f343', 'f347'] +
    expand_range('f', 351, 358) + ['f359']+['f103']
)
# === 3. Imputation Dictionary ===
imputation_values = {}

# === 4. Helper ===
def get_related_columns(col, df_columns):
    related = []
    if col in df_columns:
        related.append(col)
    log_col = f"{col}_log"
    if log_col in df_columns:
        related.append(log_col)
    return related

# === 5. Impute with ZERO ===
for col in impute_with_zero_cols:
    for related_col in get_related_columns(col, df.columns):
        df[related_col].fillna(0, inplace=True)
        imputation_values[related_col] = 0

# === 6. Impute with MEDIAN or MODE ===
for col in impute_with_median_or_mode_cols:
    for related_col in get_related_columns(col, df.columns):
        if pd.api.types.is_numeric_dtype(df[related_col]):
            value = df[related_col].median()
        else:
            mode_val = df[related_col].mode()
            value = mode_val[0] if not mode_val.empty else None

        if value is not None:
            df[related_col].fillna(value, inplace=True)
            imputation_values[related_col] = value

# === 7. Save Imputation Values ===
imputation_df = pd.DataFrame.from_dict(imputation_values, orient='index', columns=['impute_value'])
imputation_df.index.name = 'column'
imputation_df.reset_index(inplace=True)

# You can now save it to CSV or reuse it later
imputation_df.to_csv('imputation_values.csv', index=False)

print("✅ Final imputation complete. All NaNs should now be handled.")
print(f"🔍 Imputation values saved for {len(imputation_values)} columns.")
```

```python
import pandas as pd
import numpy as np

# --- Prerequisite: Ensure your DataFrame 'df' has the necessary columns ---
# This includes the remaining one-hot encoded offer categories (f227, f228, etc.),
# and the historical spend columns for both 30 and 180 days.

print("--- Creating Spend-to-Category Relevance Ratio Features ---")

# --- 1. Define Mappings and Column Lists ---

# Dictionary for 30-day spend features
category_to_spend_col_last30 = {
    'f232': 'f152_log',  # Travel Offer -> Airline Spend (30d)
    'f228': 'f153_log',  # Entertainment Offer -> Entertainment Spend (30d)
    'f227': 'f157_log',  # Dining Offer -> Restaurant Spend (30d)
    'f231': 'f158_log'   # Shopping Offer -> Retail Spend (30d)
}

# Dictionary for 180-day spend features
category_to_spend_col_last180 = {
    'f232': 'f153_log',  # Travel Offer -> Airline Spend (180d)
    'f228': 'f164_log',  # Entertainment Offer -> Entertainment Spend (180d)
    'f231': 'f169_log',  # Shopping Offer -> Retail Spend (180d)
}

# Lists of all spend columns to calculate totals for each period
all_spend_cols_last_30 = ['f152_log', 'f153_log', 'f155_log', 'f156_log', 'f157_log', 'f158_log', 'f159_log', 'f160_log', 'f161_log', 'f162_log']
all_spend_cols_last_180 = ['f163_log', 'f164_log', 'f165_log', 'f166_log', 'f167_log', 'f168_log', 'f169_log', 'f170_log', 'f171_log', 'f172_log', 'f173_log']

# --- 2. Calculate Total Historical Spend for Each Period ---
df['total_hist_spend_30'] = df[all_spend_cols_last_30].sum(axis=1)
df['total_hist_spend_180'] = df[all_spend_cols_last_180].sum(axis=1)

# --- 3. Calculate the Relevance Ratios ---
# Initialize the new feature columns
df['spend_to_category_match_30'] = 0.0
df['spend_to_category_match_180'] = 0.0

# Loop for the 30-day ratio
for category_ohe, spend_col in category_to_spend_col_last30.items():
    if category_ohe in df.columns: # Check if the category column exists
        is_category_active = df[category_ohe] == 1
        # For active rows, calculate the ratio of specific spend to total 30-day spend
        df.loc[is_category_active, 'spend_to_category_match_30'] = \
            df.loc[is_category_active, spend_col] / (df.loc[is_category_active, 'total_hist_spend_30'] + 1)

# Loop for the 180-day ratio
for category_ohe, spend_col in category_to_spend_col_last180.items():
    if category_ohe in df.columns: # Check if the category column exists
        is_category_active = df[category_ohe] == 1
        # For active rows, calculate the ratio of specific spend to total 180-day spend
        df.loc[is_category_active, 'spend_to_category_match_180'] = \
            df.loc[is_category_active, spend_col] / (df.loc[is_category_active, 'total_hist_spend_180'] + 1)

# --- 4. Clean up temporary total columns ---
df.drop(columns=['total_hist_spend_30', 'total_hist_spend_180'], inplace=True)

print("\n✅ 'spend_to_category_match_30' and 'spend_to_category_match_180' features created.")
print("\nSample of new features:")
print(df[['spend_to_category_match_30', 'spend_to_category_match_180']].head())
```

```python
import pandas as pd
import numpy as np
from tqdm.autonotebook import tqdm

# --- Prerequisites ---
# This code assumes you have:
# 1. Your main preprocessed DataFrame, 'df', with a unique row identifier 'id1', 
#    the offer ID 'id3', and the event timestamp 'datetime'.
# 2. A new DataFrame, 'events_df', loaded from your events file, with columns 
#    'id3', 'id4' (Impression Timestamp), and 'id7' (Click Timestamp).

print("--- Starting Offer Velocity Feature Creation (Highly Memory-Efficient Method) ---")

# --- Step 1: Prepare the Event Streams ---
# a. Prepare the offer impression events
impressions_df = events_df[['id3', 'id4']].copy()
impressions_df.rename(columns={'id4': 'timestamp'}, inplace=True)
impressions_df['is_impression'] = 1
impressions_df['is_click'] = 0

# b. Prepare the offer click events
clicks_df = events_df[['id3', 'id7']].dropna().copy()
clicks_df.rename(columns={'id7': 'timestamp'}, inplace=True)
clicks_df['is_impression'] = 0
clicks_df['is_click'] = 1

# c. Combine ONLY offer-related events into a single log
offer_events_log = pd.concat([impressions_df, clicks_df], ignore_index=True)
offer_events_log['timestamp'] = pd.to_datetime(offer_events_log['timestamp'])
offer_events_log.sort_values('timestamp', inplace=True)
print(f"Combined {len(offer_events_log)} offer-specific events into a single log.")

# d. Prepare the main user observation dataframe
observations_df = df[['id1', 'id3', 'datetime']].copy()
observations_df.rename(columns={'datetime': 'timestamp'}, inplace=True)
observations_df['timestamp'] = pd.to_datetime(observations_df['timestamp'])
observations_df.sort_values('timestamp', inplace=True)
print(f"Prepared {len(observations_df)} user observation events.")

# --- Step 2: Process Each Offer Individually to Save Memory ---
print("Processing each offer's history individually to conserve memory...")
all_merged_features = []
unique_offers = offer_events_log['id3'].unique()

for offer_id in tqdm(unique_offers):
    # Filter the logs for only the current offer
    single_offer_events = offer_events_log[offer_events_log['id3'] == offer_id].copy()
    single_offer_observations = observations_df[observations_df['id3'] == offer_id].copy()

    # Skip if there are no user observations for this offer
    if single_offer_observations.empty:
        continue

    # Set timestamp as index for rolling calculation
    single_offer_events.set_index('timestamp', inplace=True)

    # Calculate rolling features for this single offer
    windows = ['1h', '3h', '6h', '9h', '24h', '3d']
    for window in windows:
        rolling_sum = single_offer_events[['is_impression', 'is_click']].rolling(window).sum()
        
        # Shift to prevent data leakage (no groupby needed as we have only one offer)
        shifted_rolling_sum = rolling_sum.shift(1)
        
        imp_col_name = f'offer_impressions_last_{window}'
        click_col_name = f'offer_clicks_last_{window}'
        
        single_offer_events[imp_col_name] = shifted_rolling_sum['is_impression']
        single_offer_events[click_col_name] = shifted_rolling_sum['is_click']

    single_offer_events.reset_index(inplace=True)

    # Perform the memory-efficient merge_asof for this single offer
    merged_offer_df = pd.merge_asof(
        single_offer_observations,
        single_offer_events,
        on='timestamp',
        by='id3',
        direction='backward'
    )
    all_merged_features.append(merged_offer_df)

# --- Step 3: Combine Results and Merge Back to Main DataFrame ---
print("Combining results from all offers...")

# THIS IS THE FIX: Add a check to ensure the list is not empty before concatenating.
if not all_merged_features:
    raise ValueError("Processing failed. The 'all_merged_features' list is empty. "
                     "This likely means that none of the offer IDs ('id3') in your "
                     "'events_df' matched any of the offer IDs in your main 'df'. "
                     "Please check for data type mismatches or discrepancies in the 'id3' columns.")

final_features_df = pd.concat(all_merged_features, ignore_index=True)

# Define the list of all new feature columns created
windows = ['1h', '3h', '6h', '9h', '24h', '3d']
new_feature_cols = []
for window in windows:
    new_feature_cols.append(f'offer_impressions_last_{window}')
    new_feature_cols.append(f'offer_clicks_last_{window}')

# Select only the necessary columns for the final merge
features_to_merge = final_features_df[['id1'] + new_feature_cols]

# Merge the new features back to the original dataframe
df = df.merge(features_to_merge, on='id1', how='left')

# Fill any NaNs that might occur if an observation has no prior offer history
df[new_feature_cols] = df[new_feature_cols].fillna(0)

print("\n✅ Offer velocity features created and merged successfully.")
print("\nSample of new features added:")
print(df[new_feature_cols].head())
```

```python
import pandas as pd
import numpy as np

print("--- Starting Leak-Proof Historical Impressions Feature Creation ---")

try:
    # --- 1. Define your offer category columns ---
    offer_category_cols = [
        'f227', 'f228', 'f230', 'f231', 'f232'  # Extend as needed
    ]

    # --- 2. Basic preparation ---
    print("Preparing master log with all events...")

    # Ensure datetime is in datetime format
    df_final['datetime'] = pd.to_datetime(df_final['datetime'])

    # Sort by user and time to ensure proper leak-proof sequence
    df_final.sort_values(['id2', 'datetime'], inplace=True)

    # Copy the full df to avoid modifying the original
    master_log = df.copy()

    # --- 3. Calculate Historical Impressions Only ---
    print("Calculating historical impressions (leak-proof)...")

    new_feature_cols = []

    for category_col in offer_category_cols:
        # a. Cumulative impressions (views) — shifted to prevent leakage
        hist_views = master_log.groupby('id2')[category_col].cumsum()
        hist_views = hist_views.groupby(master_log['id2']).shift(1).fillna(0)

        # b. Define column name and store
        view_col = f'{category_col}_user_hist_impressions'
        new_feature_cols.append(view_col)

        # c. Assign to master log
        master_log[view_col] = hist_views

    print("\n✅ Leak-proof impression features created successfully.")
    print("\nSample of new features:")
    print(master_log[new_feature_cols].head())

except KeyError as e:
    print(f"\n❌ ERROR: Missing column: {e}")
```

```python
import pandas as pd
import numpy as np

print("--- Starting Concurrent Impression & Click Feature Creation (Leak-Proof, Single df) ---")

try:
    # --- Step 1: Establish cutoff time from the current df ---
    df['datetime'] = pd.to_datetime(df['datetime'])
    cutoff_time = df['datetime'].max()
    print(f"Using max timestamp in df as cutoff: {cutoff_time}")

    # --- Step 2: Prepare historical impression and click logs from events_df ---
    impressions_df = events_df[['id4']].rename(columns={'id4': 'timestamp'})
    impressions_df['is_impression'] = 1
    impressions_df['is_click'] = 0

    clicks_df = events_df[['id7']].dropna().rename(columns={'id7': 'timestamp'})
    clicks_df['is_impression'] = 0
    clicks_df['is_click'] = 1

    all_events = pd.concat([impressions_df, clicks_df], ignore_index=True)
    all_events['timestamp'] = pd.to_datetime(all_events['timestamp'])

    historical_log = all_events[all_events['timestamp'] <= cutoff_time].copy()
    historical_log.sort_values('timestamp', inplace=True)

    historical_log['cumulative_impressions'] = historical_log['is_impression'].cumsum()
    historical_log['cumulative_clicks'] = historical_log['is_click'].cumsum()

    historical_log['concurrent_impressions'] = historical_log['cumulative_impressions'].shift(1).fillna(0)
    historical_log['concurrent_clicks'] = historical_log['cumulative_clicks'].shift(1).fillna(0)

    # Drop duplicates to keep latest historical state at each timestamp
    historical_timestamp_counts = historical_log[['timestamp', 'concurrent_impressions', 'concurrent_clicks']]
    historical_timestamp_counts = historical_timestamp_counts.drop_duplicates(subset='timestamp', keep='last')
    historical_timestamp_counts.rename(columns={'timestamp': 'datetime'}, inplace=True)

    print(f"Prepared {len(historical_timestamp_counts)} historical timestamps with impression & click counts.")

    # --- Step 3: Merge into df ---
    df.sort_values('datetime', inplace=True)

    df = pd.merge_asof(
        df,
        historical_timestamp_counts,
        on='datetime',
        direction='backward'
    )

    df[['concurrent_impressions', 'concurrent_clicks']] = df[['concurrent_impressions', 'concurrent_clicks']].fillna(0)

    print("\n✅ Concurrent impression & click features added to df successfully.")
    print(df[['datetime', 'concurrent_impressions', 'concurrent_clicks']].head())

except KeyError as e:
    print(f"\n❌ ERROR: Missing column: {e}")
```

```python
import pandas as pd
import numpy as np
from tqdm.autonotebook import tqdm

print("--- Starting Placement Historical CTR Feature Creation (Leak-Proof for Full df) ---")

try:
    # --- Step 1: Setup ---
    df['datetime'] = pd.to_datetime(df['datetime'])
    events_df['id4'] = pd.to_datetime(events_df['id4'])
    events_df['id7'] = pd.to_datetime(events_df['id7'])

    cutoff_time = df['datetime'].max()
    print(f"Using cutoff time based on df: {cutoff_time}")

    # --- Step 2: Build Placement Mapping (id3 -> id6) from events_df up to cutoff ---
    print("Building placement ID map...")
    events_df['id3'] = events_df['id3'].astype(str)
    valid_events = events_df[events_df['id4'] <= cutoff_time].copy()

    id3_to_id6_map = valid_events.groupby('id3')['id6'].agg(lambda x: x.mode().iloc[0] if not x.mode().empty else np.nan).reset_index()
    print("Placement ID map complete.")

    # --- Step 3: Prepare historical events for CTR ---
    impressions_df = events_df[['id6', 'id4']].rename(columns={'id4': 'timestamp'})
    impressions_df['is_impression'] = 1
    impressions_df['is_click'] = 0

    clicks_df = events_df[['id6', 'id7']].dropna().rename(columns={'id7': 'timestamp'})
    clicks_df['is_impression'] = 0
    clicks_df['is_click'] = 1

    placement_events = pd.concat([impressions_df, clicks_df], ignore_index=True)
    placement_events = placement_events[placement_events['timestamp'] <= cutoff_time].copy()
    placement_events.dropna(subset=['timestamp'], inplace=True)
    placement_events.sort_values(['id6', 'timestamp'], inplace=True)

    print(f"Prepared {len(placement_events)} placement events for CTR calculation.")

    # --- Step 4: Calculate Cumulative Impressions & Clicks ---
    placement_events['hist_impressions'] = placement_events.groupby('id6')['is_impression'].cumsum()
    placement_events['hist_clicks'] = placement_events.groupby('id6')['is_click'].cumsum()

    placement_events['hist_impressions'] = placement_events.groupby('id6')['hist_impressions'].shift(1).fillna(0)
    placement_events['hist_clicks'] = placement_events.groupby('id6')['hist_clicks'].shift(1).fillna(0)

    placement_events['placement_historical_ctr'] = placement_events['hist_clicks'] / (placement_events['hist_impressions'] + 1)

    ctr_lookup = placement_events[['id6', 'timestamp', 'placement_historical_ctr']].drop_duplicates(
        subset=['id6', 'timestamp'], keep='last'
    ).copy()

    print("CTR lookup table ready.")

    # --- Step 5: Merge CTR into full df ---
    def add_placement_ctr_feature_loop(df, id_map, ctr_lookup):
        df['id3'] = df['id3'].astype(str)
        df = pd.merge(df, id_map, on='id3', how='left')
        df['id6'] = df['id6'].fillna('unknown_placement')

        all_parts = []
        for placement_id in tqdm(df['id6'].unique(), desc="Merging CTR by placement"):
            df_part = df[df['id6'] == placement_id].copy()
            ctr_part = ctr_lookup[ctr_lookup['id6'] == placement_id].copy()

            df_part.sort_values('datetime', inplace=True)
            ctr_part.sort_values('timestamp', inplace=True)

            merged = pd.merge_asof(
                df_part,
                ctr_part.drop(columns=['id6']),
                left_on='datetime',
                right_on='timestamp',
                direction='backward'
            )
            all_parts.append(merged)

        full_df = pd.concat(all_parts, ignore_index=True)
        full_df['placement_historical_ctr'].fillna(0, inplace=True)
        return full_df

    df_final = add_placement_ctr_feature_loop(df, id3_to_id6_map, ctr_lookup)

    print("\n✅ Placement historical CTR added successfully.")
    print(df_final[['datetime', 'id3', 'placement_historical_ctr']].head())

except KeyError as e:
    print(f"\n❌ ERROR: Missing column: {e}")
except Exception as e:
    print(f"\n❌ Unexpected error: {e}")
```

```python
import pandas as pd
import numpy as np
import warnings
from tqdm.auto import tqdm
from sentence_transformers import SentenceTransformer
from sklearn.decomposition import PCA

# --- Housekeeping ---
warnings.filterwarnings("ignore")
pd.options.mode.chained_assignment = None
tqdm.pandas()

print("🚀 Starting feature creation on df_final")

# --- 1. Prepare Data ---
df = df_final.copy()
offers_df['id3'] = offers_df['id3'].astype(str)
df['id3'] = df['id3'].astype(str)
df = pd.merge(df, offers_df[['id3', 'f378']], on='id3', how='left')

def clean_label(txt: str):
    if not isinstance(txt, str) or not txt.strip():
        return '__none__'
    s = txt.lower().replace('&',' ').replace('/',' ').replace('-',' ')
    s = ' '.join(s.split())
    if len(s) == 0:
        return '__none__'
    pct = sum(c.isalpha() for c in s) / len(s)
    return '__hash__' if pct < 0.2 else s

df['label_clean'] = df['f378'].map(clean_label)

df['id4'] = pd.to_datetime(df['id4'], errors='coerce')
df.sort_values('id4', inplace=True)

# --- 2. Train SentenceTransformer + PCA ---
all_text = df['label_clean'].unique()
all_text = [t for t in all_text if t not in ['__hash__', '__none__']]

print(f"Found {len(all_text)} unique cleaned texts for embedding.")
n_components = min(32, len(all_text))
print(f"Setting PCA n_components to: {n_components}")

vec_dict = {}
if n_components > 0:
    print("Embedding and reducing offer text...")
    st_model = SentenceTransformer('all-MiniLM-L6-v2')
    raw_vecs = st_model.encode(all_text, batch_size=128,
                               show_progress_bar=True,
                               normalize_embeddings=True)

    pca = PCA(n_components=n_components, random_state=42).fit(raw_vecs)
    transformed = pca.transform(raw_vecs)

    vec_dict = {txt: v for txt, v in zip(all_text, transformed)}
    vec_dict['__hash__'] = transformed.mean(axis=0)
    vec_dict['__none__'] = np.zeros(n_components)
else:
    print("⚠️  Not enough valid texts → falling back to zero embeddings.")

# --- 3. Lookup embeddings ---
def lookup(label):
    if not vec_dict:
        return np.zeros(n_components if n_components > 0 else 32)
    return vec_dict.get(label, vec_dict['__hash__'])

df['vector'] = df['label_clean'].progress_apply(lookup)
vec_matrix = np.stack(df['vector'].values)
pca_cols = [f'pca_{i}' for i in range(vec_matrix.shape[1])]
df[pca_cols] = vec_matrix

# --- 4. Session Similarity Features ---
print("Calculating session-level similarity features...")
df.sort_values(['id2', 'id4'], inplace=True)
df.reset_index(drop=True, inplace=True)

vec_matrix = np.stack(df['vector'].values)
y_values = df['y'].values
id2_values = df['id2'].values

# Similarity to last clicked offer
last_vec = np.zeros_like(vec_matrix)
last_seen = {}
for i in tqdm(range(len(df)), desc="→ sim_to_last_click"):
    uid = id2_values[i]
    vec = vec_matrix[i]
    y = y_values[i]
    
    if uid in last_seen:
        last_vec[i] = last_seen[uid]
    if y == 1:
        last_seen[uid] = vec
        
df['sim_to_last_click'] = np.einsum('ij,ij->i', vec_matrix, last_vec) / (
    np.linalg.norm(vec_matrix, axis=1) *
    np.linalg.norm(last_vec, axis=1) + 1e-12)

# Similarity to running session centroid
centroid = np.zeros_like(vec_matrix)
sum_vec = {}
count = {}
for i in tqdm(range(len(df)), desc="→ sim_to_session_centroid"):
    uid = id2_values[i]
    vec = vec_matrix[i]

    if uid in sum_vec:
        centroid[i] = sum_vec[uid] / count[uid]
        sum_vec[uid] += vec
        count[uid] += 1
    else:
        centroid[i] = np.zeros_like(vec)
        sum_vec[uid] = vec.copy()
        count[uid] = 1

df['sim_to_session_centroid'] = np.einsum('ij,ij->i', vec_matrix, centroid) / (
    np.linalg.norm(vec_matrix, axis=1) *
    np.linalg.norm(centroid, axis=1) + 1e-12)

# --- 5. Finalize ---
df.drop(columns=['f378', 'label_clean', 'vector'], inplace=True)
df.fillna(0, inplace=True)

df_final_with_embeddings = df
new_feature_count = len(pca_cols) + 2
print(f"\n✅ Added {new_feature_count} new features to df_final.")
print(f"df_final shape: {df_final_with_embeddings.shape}")
```

```python




# --- 3. Create Industry-Level Transactional Features (Leak-Proof) ---
print("\n--- Creating Industry-Level Transactional Features ---")

# a. Determine the end time of the training period
train_df['datetime'] = pd.to_datetime(train_df['datetime'])
train_end_time = train_df['datetime'].max()
print(f"Training data ends at: {train_end_time}")

# b. Prepare and filter the transaction data
transaction_df['f370'] = pd.to_datetime(transaction_df['f370'])
historical_transaction_df = transaction_df[transaction_df['f370'] <= train_end_time].copy()
print(f"Created historical transaction log with {len(historical_transaction_df)} transactions.")

# c. Calculate aggregate stats from the historical transaction data
print("Calculating aggregate stats for each industry from historical data...")
industry_avg_spend = historical_transaction_df.groupby('id8')['f367'].mean().rename('industry_avg_transaction_amount')
industry_transaction_count = historical_transaction_df.groupby('id8').size().rename('industry_transaction_count')

# d. Calculate the debit-to-credit ratio for each industry
debit_credit_counts = historical_transaction_df.groupby('id8')['f369'].value_counts().unstack().fillna(0)
debit_credit_counts['industry_debit_credit_ratio'] = \
    debit_credit_counts.get('D', 0) / (debit_credit_counts.get('C', 0) + 1)

# e. Combine all stats into a single DataFrame
industry_stats_df = pd.concat([industry_avg_spend, industry_transaction_count, debit_credit_counts['industry_debit_credit_ratio']], axis=1)
print(f"Calculated stats for {len(industry_stats_df)} unique industries.")


# f. Create a map from offer ID ('id3') to its Industry Code ('id10')
# We use drop_duplicates to handle cases where an offer might have multiple entries.
offer_to_industry_map = offerings_df.drop_duplicates(subset='id3').set_index('id3')['id10']

# --- THIS IS THE MEMORY-EFFICIENT MODIFICATION ---
# g. Define a reusable function to merge features using .map()
def add_industry_features(df_to_process, offer_map, stats_df):
    # Ensure keys are the same type
    df_to_process['id3'] = df_to_process['id3'].astype(str)
    offer_map.index = offer_map.index.astype(str)
    
    # Step 1: Add the industry code ('id10') using .map()
    df_to_process['id10'] = df_to_process['id3'].map(offer_map)
    
    # Step 2: Add the stats using .map() on the new 'id10' column
    df_to_process['id10'] = df_to_process['id10'].astype(str)
    stats_df.index = stats_df.index.astype(str)

    # Create individual maps for each stat for efficient mapping
    avg_spend_map = stats_df['industry_avg_transaction_amount']
    count_map = stats_df['industry_transaction_count']
    ratio_map = stats_df['industry_debit_credit_ratio']

    df_to_process['industry_avg_transaction_amount'] = df_to_process['id10'].map(avg_spend_map)
    df_to_process['industry_transaction_count'] = df_to_process['id10'].map(count_map)
    df_to_process['industry_debit_credit_ratio'] = df_to_process['id10'].map(ratio_map)
    
    # c. Fill missing values
    new_cols = ['industry_avg_transaction_amount', 'industry_transaction_count', 'industry_debit_credit_ratio']
    df_to_process[new_cols] = df_to_process[new_cols].fillna(0)
    
    return df_to_process

# h. Apply the feature creation to both train and validation sets
print("Applying features to training set...")
train_df = add_industry_features(train_df, offer_to_industry_map, industry_stats_df)

print("Applying features to validation set...")
val_df = add_industry_features(val_df, offer_to_industry_map, industry_stats_df)
print("✅ Industry features created and merged successfully.")

```

```python
import pandas as pd
import numpy as np
import xgboost as xgb
import gc
from tqdm.autonotebook import tqdm
from xgboost.callback import TrainingCallback

# --- Main Script ---

# Assume 'df' is your fully preprocessed DataFrame, sorted by a datetime column.
# Make sure 'y' is your target variable and the rest are features

# --- 2. Target Encode Offer ID (Leak-Proof) ---

# --- 4. Prepare Data and Create Group Arrays for LTR ---
# CRITICAL FOR RANKING: Sort data by the group ID (user ID).

train_df = train_df.sort_values('id2')
val_df = val_df.sort_values('id2')

# Define columns to drop
columns_to_drop = ['y', 'id2','id3','id1','id10', 'id4', 'id5','f378', 'datetime','id8_x','id8_y','id8','id10_y','id11_y','mystery_component']
X_train = train_df.drop(columns=columns_to_drop, errors='ignore')
y_train = train_df['y']
X_val = val_df.drop(columns=columns_to_drop, errors='ignore')
y_val = val_df['y']

# Ensure column order is the same
X_val = X_val[X_train.columns]
features = X_train.columns.tolist()

# Create the group arrays by counting items per user in the sorted data
print("--- Creating group arrays for DMatrix ---")
train_groups = train_df.groupby('id2').size().to_numpy()
val_groups = val_df.groupby('id2').size().to_numpy()


# --- 5. Convert to DMatrix with Group Info and Define Parameters ---
print("--- Converting data to DMatrix format for LTR ---")
dtrain = xgb.DMatrix(X_train, label=y_train, group=train_groups, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, group=val_groups, feature_names=features)

# Clean up memory

# Use updated parameters for pairwise ranking
print("--- Using Learning-to-Rank parameters ---")
xgb_params_tuned = {
    'objective':        'rank:ndcg',
    'eval_metric':      'map@7',
    
    # — Reduce learning rate for smoother convergence —
    'eta':              0.03,      
    
    # — Shallower trees to limit model complexity —
    'max_depth':        6,         

    # — Sample only a fraction of rows & cols per tree to reduce variance —
    'subsample':        0.8,
    'colsample_bytree': 0.8,

    # — Require more weight in a child before splitting (pre‑pruning) —
    'min_child_weight': 10,        

    # — Add a little regularization on leaf weights —
    'gamma':            1,          # minimum loss reduction to make a split
    'reg_alpha':        0.5,        # L1 regularization
    'reg_lambda':       2.0,        # L2 regularization

    'tree_method':      'hist',
    'seed':             42,
}

watchlist = [(dtrain, 'train'), (dval, 'val')]
n_estimators = 1000

# Custom callback for a tqdm progress bar
class TqdmCallback(TrainingCallback):
    def __init__(self, total):
        self.pbar = tqdm(total=total)
    def after_iteration(self, model, epoch, evals_log):
        self.pbar.update(1)
        return False
    def after_training(self, model):
        self.pbar.close()
        return model

# --- 6. Train the Model ---
print("\n--- Training LTR Model ---")
model = xgb.train(
    params=xgb_params_tuned,
    dtrain=dtrain,
    num_boost_round=n_estimators,
    evals=watchlist,
    verbose_eval=10,
    early_stopping_rounds=50,
    callbacks=[TqdmCallback(total=n_estimators)]
)

# --- 7. Evaluate and Save Model ---
print("\n--- Training Complete ---")
print(f"Best Validation MAP@7: {model.best_score:.6f} at iteration {model.best_iteration}")
model_path = 'xgboost_ltr_model.json'
model.save_model(model_path)
print(f"✅ Model saved to: {model_path}")
```
